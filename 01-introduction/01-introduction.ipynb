## 2. What is AI?

**Artificial Intelligence (AI)** is a vast field, but at its core, it is the capacity of machines to learn, perceive, and decide which paths to follow, rationally, in the face of certain situations.

We can categorize AI approaches into two main groups:

1.  **Symbolic (Logic-Based) AI:** Relies on logical reasoning and the manipulation of data (letters, numbers, words) to represent knowledge and solve problems. This was the dominant paradigm in early AI.
2.  **Connectionist AI (Neural Networks):** Simulates the functioning of biological neurons. These systems *learn* from data and adapt over time. This is the foundation for modern Deep Learning.

## 3. A Very Brief History of AI

The idea of AI has been around for a long time, but the field formally began in the 20th century.

- **1940s-1950s (The Dawn):** **Alan Turing** introduces the "Turing Test" (or "Imitation Game") to determine if a machine can exhibit intelligent behavior indistinguishable from a human's.
- **1997 (A Milestone):** IBM's **Deep Blue** computer defeats world chess champion Garry Kasparov. This was a major victory for "brute force" search and symbolic AI.
- **2010s-Present (The Deep Learning Boom):** Fueled by massive amounts of data and powerful computers (GPUs), Connectionist AI (Deep Learning) achieves superhuman performance in tasks like image recognition, natural language processing, and more.

### Puzzle 3 (Numeric Sequences)

![Puzzle3](./assets/puzzles/puzzle3.png)

**Task:** Find the next number in each sequence.

**Sequence A:** `2, 4, 7, 11, 16, ?`

**Sequence B:** `0, 1, 1, 2, 3, ?`

```python
# Your turn! Try to solve Puzzle 3 here.

def solve_sequence_A():
    # Hint: Look at the *difference* between consecutive numbers.
    # 4 - 2 = 2
    # 7 - 4 = 3
    # 11 - 7 = 4
    # 16 - 11 = 5
    # What's next?
    next_num = 16 + 6
    return next_num

def solve_sequence_B():
    # Hint: What is the relationship between a number and the two that came before it?
    # 0 + 1 = 1
    # 1 + 1 = 2
    # 1 + 2 = 3
    # What's next?
    next_num = 2 + 3
    return next_num

print(f"Next in Sequence A: {solve_sequence_A()}")
print(f"Next in Sequence B: {solve_sequence_B()} (This is the Fibonacci sequence)")
```
