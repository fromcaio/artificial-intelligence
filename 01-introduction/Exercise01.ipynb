Form groups of up to **4 students** and discuss the following questions. 

Appoint one person to take notes in the Markdown cell below. Be prepared to share your group's thoughts with the class.

### Discussion Points

1.  **Aplicações da IA:** What other applications of AI do you know or use, besides the ones mentioned in class (e.g., Netflix, Waze, Siri)?
    
2.  **Mitos e Verdades:** What are some common myths (or truths) you hear about AI? (e.g., "AI will steal all jobs," "AI is conscious," "AI only works for big companies").

3.  **Uso Atual:** What AI-powered systems do you *personally* use every day? (Think about virtual assistants, games, email spam filters, social media feeds, etc.).

4.  **A Grande Pergunta:** Will AI replace humans? Why or why not? (Consider what "replace" means. In all tasks? In some tasks? Will it create new human roles?)
