### Goal

Design a simple intelligent agent for a small, well-defined domain. This exercise combines the concepts from all three notebooks: defining an agent, specifying its task environment (PEAS), and choosing a simple Knowledge Representation (KR) model for it to use.

### Task

1.  **Pick a Domain:** Choose a simple domain. 
    * *Example 1:* A warehouse robot that moves items from a shelf to a packing station.
    * *Example 2:* A simple game NPC (Non-Player Character) that patrols between two points.
    * *Example 3:* A shopping assistant bot that recommends a product based on 2-3 questions.

2.  **Specify PEAS:** Define the **P**erformance, **E**nvironment, **A**ctuators, and **S**ensors for your agent.

3.  **Choose KR:** Choose a simple knowledge representation. 
    * *Warehouse Robot:* A dictionary (Frame) for `Items` and their `Locations`.
    * *Game NPC:* A `Script` for its patrol path (Scene 1: go to A, Scene 2: wait, Scene 3: go to B).
    * *Shopping Bot:* A `Semantic Network` (graph) of product categories.

4.  **Implement a Prototype:** Write a *very* simple Python program (using dictionaries or classes) that simulates your agent. A few `if/else` rules are enough. 

5.  **Demonstrate:** Print a short log showing the agent perceiving, making a decision based on its KR, and acting.

### Example: Warehouse Robot

Here is a template for the "Warehouse Robot" domain. You can fill this out or create your own.

```python

# 1. Define Domain: Warehouse Robot
print("--- DOMAIN: Warehouse Robot ---")

# 2. PEAS Specification (as a dictionary)
PEAS = {
    "Performance": ["Time to pack order", "Number of wrong items"],
    "Environment": ["Warehouse grid (10x10)", "Shelves", "Packing station", "Items"],
    "Actuators": ["Motors (move F/B/L/R)", "Gripper (pick, drop)"],
    "Sensors": ["Camera (find item)", "Position sensor (e.g., (x,y))", "Gripper sensor (holding_item?)"]
}
print("PEAS:", PEAS)

# 3. Knowledge Representation (Simple Frame/Dict)
# This is the agent's "knowledge" of the world.
knowledge_base = {
    "item_locations": {
        "item_A": "(1, 5)",
        "item_B": "(3, 8)",
        "item_C": "(9, 2)"
    },
    "packing_station": "(0, 0)",
    "rules": [
        "if not holding_item -> move toward item_location",
        "if at item_location and not holding_item -> pick",
        "if holding_item -> move toward packing_station",
        "if at packing_station and holding_item -> drop"
    ]
}
print("\nKnowledge Base:", knowledge_base)

# 4. Agent Program (Simple Agent Loop)
class WarehouseAgent:
    def __init__(self, knowledge):
        self.knowledge = knowledge
        self.current_state = {
            "position": "(0, 0)",
            "holding_item": None
        }
        self.order_list = ["item_A", "item_C"]
        self.log = []

    def get_action(self):
        """This is the agent's function f(P*) -> A"""
        if not self.order_list:
            return "NoOp" # Order complete
        
        target_item = self.order_list[0]
        target_loc = self.knowledge["item_locations"][target_item]
        packing_loc = self.knowledge["packing_station"]
        
        # Rule 3 & 4: If holding item, go to packing station
        if self.current_state["holding_item"] == target_item:
            if self.current_state["position"] == packing_loc:
                return "Drop"
            else:
                return f"Move to {packing_loc}"
        
        # Rule 1 & 2: If not holding item, go to item location
        else:
            if self.current_state["position"] == target_loc:
                return "Pick"
            else:
                return f"Move to {target_loc}"

    def simulate_step(self, action):
        """Mock environment update"""
        self.log.append(f"PERCEPT: {self.current_state} -> ACTION: {action}")
        if action == "Pick":
            self.current_state["holding_item"] = self.order_list[0]
        elif action == "Drop":
            self.current_state["holding_item"] = None
            self.order_list.pop(0) # Item is done
        elif "Move to" in action:
            self.current_state["position"] = action.split(" ")[-1]

# 5. Demonstrate
print("\n--- Agent Simulation --- ")
agent = WarehouseAgent(knowledge_base)

# Run for 6 steps
for _ in range(6):
    act = agent.get_action()
    agent.simulate_step(act)
    if act == "NoOp":
        break

for entry in agent.log:
    print(entry)
```
