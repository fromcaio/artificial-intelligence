## Heuristica de Refinamento

Vamos implementar uma solução que utilize heuristica de refinamento para o problema da mochila (maximização de beneficio).

Os dados que iremos utilizar no nosso problema são:

- Peso = {4,5,7,9,6}
- Beneficio = {2,2,3,4,4}

Solucao inicial viavel
- s = ( 0 1 1 1 0 )

Peso Maximo
- 23

```python
import numpy

peso = [4,5,7,9,6]
beneficio = [2,2,3,4,4]
solucao_inicial = [0,1,1,1,0]
capacidade_mochila = 23

def calcula_peso(solucao):
    peso_total = 0
    for i in range(len(solucao)):
        peso_total += solucao[i] * peso[i]
    return peso_total

def calcula_beneficio(solucao):
    beneficio_total = 0
    for i in range(len(solucao)):
        beneficio_total += solucao[i] * beneficio[i]
    return beneficio_total

def gera_vizinho(solucao):
    vizinho = solucao.copy()
    index = numpy.random.randint(0, len(solucao))
    vizinho[index] = 1 - vizinho[index]  # Inverte o valor (0 -> 1 ou 1 -> 0)
    return vizinho
```
